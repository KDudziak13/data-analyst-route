1.	Napisz funkcję, która przyjmuje imię i wypisuje "Cześć, [imię]!".

```python
def przywitaj():
  imie = input("Podaj swoje imię: ")
  print(f"Cześć, {imie}!")
przywitaj()
```

2.	Napisz funkcję, która oblicza kwadrat liczby.

```python
def kwadrat():
  liczba = int(input("Podaj liczbę: "))
  liczba2 = liczba ** 2
  print("Kwadrat liczby to:", liczba2)
kwadrat()
```

3. Napisz funkcję, która sprawdza, czy liczba jest parzysta.

```python
def parzysta():
  liczba = int(input("Podaj liczbę: "))
  if liczba % 2 == 0:
    print("Liczba jest parzysta")
  else:
    print("Liczba nie jest parzysta")

parzysta()
```

4. Napisz funkcję, która zwraca największą liczbę z listy.

```python
def najwieksza():
  dane = input("Podaj liczby oddzielone przecinkami: ")
  lista = [int(x.strip()) for x in dane.split(",")]
  print(max(lista))
najwieksza()
```

5. Napisz funkcję, która przyjmuje listę liczb i zwraca nową listę z tylko liczbami większymi od średniej.

```python
def srednia_listy():
  dane = input("Podaj liczby oddzielone przecinkami: ")
  lista = [int(x.strip()) for x in dane.split(",")]
  srednia = sum(lista) / len(lista)
  for i in lista:
    if i > srednia:
      print(i)

srednia_listy()
```

6. Napisz funkcję, która liczy silnię liczby n (rekurencja lub pętla).

```python
import math
def silnia():
  liczba = int(input("Podaj liczbę: "))
  print(math.factorial(liczba))
silnia()
```

```python
def silnia(n):
  if n == 0 or n==1:
    return 1
  return n*silnia(n-1)


n = int(input("Podaj liczbę: "))
silnia(n)
```

```python
def silnia():
  n = int(input("Podaj liczbę: "))
  wynik = 1
  for i in range(1, n + 1):
    wynik *= i
  return wynik

silnia()
```
